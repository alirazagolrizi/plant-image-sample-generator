# Generate Sample Plant Image Folders | ساخت پوشه و تصویر نمونهٔ گیاه

Create sample train/validation class folders with generated random images.

Open the repository README for prerequisites, input assets, limitations, and run instructions. | برای پیش‌نیازها، ورودی‌ها و راهنمای اجرا README مخزن را ببینید.

```python
import os
import numpy as np
from PIL import Image

# ساختار پوشه‌ها
folders = [
    "PlantVillage/train/healthy",
    "PlantVillage/train/diseased",
    "PlantVillage/val/healthy",
    "PlantVillage/val/diseased"
]

for folder in folders:
    os.makedirs(folder, exist_ok=True)

# ساخت چند تصویر نمونه
def create_sample_image(path):
    img = Image.fromarray(np.random.randint(0, 255, (128,128,3), dtype=np.uint8))
    img.save(path)

# ساخت تصاویر نمونه
for i in range(5):
    create_sample_image(f"PlantVillage/train/healthy/sample_{i}.jpg")
    create_sample_image(f"PlantVillage/train/diseased/sample_{i}.jpg")
    create_sample_image(f"PlantVillage/val/healthy/sample_{i}.jpg")
    create_sample_image(f"PlantVillage/val/diseased/sample_{i}.jpg")

print("پوشه‌ها و تصاویر نمونه ساخته شدند!")
```
